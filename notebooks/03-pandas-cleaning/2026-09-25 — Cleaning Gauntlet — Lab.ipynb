## Lab 05 — Cleaning Gauntlet

Three hundred rows, generated messy. This is the first dataset in the course you cannot eyeball, which means you have to work from counts and assertions rather than from looking at the table and deciding it seems fine.

Deliverables: a clean frame, a decision log, a set of assertions that pass, and one business number at the end — revenue by category — that you would be willing to defend.

Keep the log as you go. Reconstructing it afterward is much harder than writing one line per step, and the write-up at the end depends on it.

### The log

Run this first, then call `log(...)` after each cleaning step.

````python
DECISIONS = []

def log(step, decision, rows_affected):
    DECISIONS.append({'step': step, 'decision': decision, 'rows': rows_affected})
    print(f'[{step}] {decision} ({rows_affected} row(s))')
````

````python
import pandas as pd, numpy as np
from io import StringIO
rng = np.random.default_rng(5)
items = ['Cheeseburger','cheese burger','Foam Finger','foam finger','Rain Poncho','rain poncho']
cats = ['Food','food','Merch','Apparel','RainGear','rain-gear']
rows = []
for i in range(300):
    rows.append({
        'order_id': i,
        'item': rng.choice(items),
        'category': rng.choice(cats),
        'qty': rng.choice([1,2,3,-1,np.nan], p=[.5,.25,.15,.05,.05]),
        'price': rng.choice(['$7.50','7.5','$12.00','24','6.0']),
    })
df = pd.DataFrame(rows)
df = pd.concat([df, df.sample(15, random_state=1)])  # inject dupes
df.head()
````

### TODO 1 — drop duplicates

````python
before = len(df)
df = df.drop_duplicates()
log('drop duplicates', 'Dropped duplicate rows', before - len(df))
````

I removed duplicate rows so the same order would not be counted more than once. This prevents duplicates from incorrectly affecting the results.

### TODO 2 — clean `price` -> float

````python
df['price'] = df['price'].str.replace('$', '', regex=False).astype(float)
log('clean price', 'Removed $ and converted price to float', len(df))
````

I removed the dollar signs from the price values and converted the column to floats. This makes the prices numeric so they can be used in calculations later.

### TODO 3 — `qty` -> numeric, drop rows with missing/negative qty

````python
df['qty'] = pd.to_numeric(df['qty'])
before = len(df)
df = df.dropna(subset=['qty'])
df = df[df['qty'] >= 0]
log('clean qty', 'Dropped missing and negative quantities', before - len(df))
````

I converted the quantity column to numeric values and removed rows with missing or negative quantities. These rows would not represent valid quantities for calculating revenue.

### TODO 4 — canonicalize `item`

Six spellings, three real products. Start by listing what you actually have, then build the mapping from that list rather than from memory.

```python
print(df['item'].value_counts())
ITEM_MAP = {...}
```

````python
print(df['item'].value_counts())

ITEM_MAP = {
    'Cheeseburger': 'Cheeseburger',
    'cheese burger': 'Cheeseburger',
    'Foam Finger': 'Foam Finger',
    'foam finger': 'Foam Finger',
    'Rain Poncho': 'Rain Poncho',
    'rain poncho': 'Rain Poncho'
}

df['item'] = df['item'].map(ITEM_MAP)
log('canonicalize item', 'Collapsed six item names into three', len(df))
````

I combined the six different item spellings into three consistent product names. This makes sure different spellings and capitalization of the same product are treated as one item.

### TODO 5 — normalize `category`

Same approach. Note that `Apparel` and `Merch` are a business decision, not a string problem — decide and log it.

````python
CATEGORY_MAP = {
    'Food': 'Food',
    'food': 'Food',
    'Merch': 'Merch',
    'Apparel': 'Merch',
    'RainGear': 'RainGear',
    'rain-gear': 'RainGear'
}

df['category'] = df['category'].map(CATEGORY_MAP)
log('normalize category', 'Combined category variants and treated Apparel as Merch', len(df))
````

I standardized the category names and chose to combine Apparel with Merch. I made this choice because apparel can reasonably be considered a type of merchandise.

### TODO 6 — prove it's clean

**TODO:** uncomment these and add two more assertions of your own — one about the item names and one about the categories.

````python
assert df.duplicated().sum() == 0
assert df['qty'].min() >= 1
assert df['price'].dtype == float
assert df['item'].nunique() == 3
assert df['category'].nunique() == 3

print('clean:', df.shape)
````

### TODO 7 — the number you would report

**TODO:** add a `revenue` column, then print revenue by category, highest first, plus the overall total. Round money to two decimals.

Then, in one sentence, state what you would tell a vendor to stock more of.

````python
df['revenue'] = df['qty'] * df['price']

revenue_by_category = df.groupby('category')['revenue'].sum().sort_values(ascending=False).round(2)

print(revenue_by_category)
print('Overall total:', round(df['revenue'].sum(), 2))
````

I calculated revenue by multiplying quantity by price, then totaled the revenue for each category. I sorted the categories from highest to lowest to see which category generated the most revenue.

I would tell the vendor to stock more Food because it generated the highest total revenue at $1,656.00.
